# Notebook 05: Behavioural Fraud Indicators

Inspect and quantify the indicators used as transaction context.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.behaviour_detector import BEHAVIOURAL_COLUMNS

transactions = pd.read_csv(PROJECT_ROOT / "data" / "synthetic" / "transactions.csv")
binary_columns = [column for column in BEHAVIOURAL_COLUMNS if column != "amount"]
display(transactions.groupby("is_fraudulent")[BEHAVIOURAL_COLUMNS].mean().round(3))
display(transactions.groupby("is_fraudulent")[binary_columns].mean().mul(100).round(1).add_suffix("%"))
```

```python
def behavioural_indicators(row):
    return {
        "suspicious_transaction_context": bool(row["is_new_recipient"] or row["recent_suspicious_sms"]),
        "otp_request": bool(row["otp_request"]),
        "new_device": bool(row["new_device"]),
    }

display(transactions.head(10).apply(behavioural_indicators, axis=1).to_frame("indicators"))
```
